```python
#EXERCÍCIO 10
nota = float(input("digite a sua nota escolar: "))
if 0.0 <= nota <= 5.4:
  print("Sinto muito, você está reprovado(a)")
elif 5.5 <= nota <= 7.4:
  print("Infelizmente, você ficou na recuperação.")
elif 7.5 <= nota <= 10.0:
  print("Parabéns! você foi aprovado(a)!")
else:
  print("nota impossível.")
```
